```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "месяц": ["Янв", "Фев", "Мар", "Апр", "Май", "Июн"],
    "заказы": [120, 135, 160, 150, 190, 230],
    "отмены": [10, 12, 9, 15, 11, 14],
})
df["доля_отмен_%"] = (df["отмены"] / df["заказы"] * 100).round(1)
df
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df["месяц"], df["заказы"], marker="o", label="Заказы")
plt.bar(df["месяц"], df["отмены"], alpha=0.5, label="Отмены")
plt.title("Заказы и отмены по месяцам")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
